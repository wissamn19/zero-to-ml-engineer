## 1. Basic Arithmetic Operations

Python can perform common mathematical operations using arithmetic operators.

### Arithmetic operators

| Operator | Meaning | Example |
|---|---|---|
| `+` | Addition | `5 + 3` → `8` |
| `-` | Subtraction | `5 - 3` → `2` |
| `*` | Multiplication | `5 * 3` → `15` |
| `/` | Division | `5 / 2` → `2.5` |
| `//` | Floor division | `5 // 2` → `2` |
| `%` | Modulo / remainder | `5 % 2` → `1` |
| `**` | Power | `5 ** 2` → `25` |

```python
a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor division:", a // b)
print("Remainder:", a % b)
print("Power:", a ** b)
```

### Example: Calculate the area of a rectangle

```python
length = 8
width = 5

area = length * width
print("Area:", area)
```

## 2. Control and Conditional Structures

Conditional statements allow a program to make decisions.

The main keywords are:
- `if`: runs code when a condition is true.
- `elif` : checks another condition if the previous one was false.
- `else` : runs when none of the previous conditions are true.

```python
age = 20

if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")
```

### Example: Grade classification

```python
grade = 75

if grade >= 90:
    print("Excellent")
elif grade >= 70:
    print("Good")
elif grade >= 50:
    print("Pass")
else:
    print("Fail")
```

### Comparison and logical operators

Common comparison operators are `==`, `!=`, `>`, `<`, `>=`, and `<=`.

Logical operators include `and`, `or`, and `not`.

```python
temperature = 25
is_sunny = True

if temperature > 20 and is_sunny:
    print("It is a good day to go outside!")
```

## 3. Looping : `for` and `while` Loops

Loops repeat a block of code multiple times.

### `for` loop

A `for` loop is commonly used when you want to repeat something for each item in a sequence or a specific number of times.

```python
for number in range(1, 6):
    print(number)
```

### Example: Loop through a list

```python
fruits = ["apple", "banana", "orange"]

for fruit in fruits:
    print(fruit)
```

### `while` loop

A `while` loop keeps running as long as its condition is true.

```python
counter = 1

while counter <= 5:
    print(counter)
    counter += 1
```

 **Important:** Make sure the condition of a `while` loop eventually becomes false. Otherwise, you can create an infinite loop.

## 4. User Input

The `input()` function allows a Python program to receive information from the user.

```python
name = input("Enter your name: ")
print("Hello", name + "!")
```

### Important: `input()` returns a string

Even if the user enters a number, `input()` initially stores it as a string. We can use typecasting to convert it into an integer or float.

```python
age = input("Enter your age: ")
age = int(age)

print("Next year you will be", age + 1)
```

## 5. Strings

A string is a sequence of characters. Strings are written using single quotes `' '` or double quotes `" "`.

```python
name = "Wissem"
message = 'Hello, Python!'

print(name)
print(message)
print(type(name))
```

### Common string operations

You can concatenate strings with `+`, repeat them with `*`, and access individual characters using indexing.

```python
first_name = "Python"
last_name = "Developer"

full_name = first_name + " " + last_name

print(full_name)
print("First character:", full_name[0])
print("Length:", len(full_name))
print("Uppercase:", full_name.upper())
print("Lowercase:", full_name.lower())
```

### f-strings

f-strings make it easy to insert variables into strings.

```python
name = "Alex"
age = 21

print(f"My name is {name} and I am {age} years old.")
```

## 6. Integers

An integer (`int`) is a whole number without a decimal part.

```python
age = 21
score = 100
temperature = -5

print(age)
print(type(age))
print(score + 50)
print(temperature * 2)
```

## 7. Floating Point Values

A floating-point number (`float`) is a number that can contain a decimal part.

```python
price = 19.99
height = 1.68
average = 15.5

print(price)
print(type(price))
print(price * 2)
```

### Integer vs float

Division with `/` normally produces a float, even when the result is mathematically a whole number.

```python
a = 10
b = 2

result = a / b

print(result)
print(type(result))
```

## 8. Typecasting

Typecasting means converting a value from one data type to another.

Common functions include:
- `int()` → converts to an integer
- `float()` → converts to a floating-point number
- `str()` → converts to a string
- `bool()` → converts to a Boolean value

```python
number_string = "25"
number_integer = int(number_string)

decimal_string = "12.5"
number_float = float(decimal_string)

number = 100
number_as_string = str(number)

print(number_integer, type(number_integer))
print(number_float, type(number_float))
print(number_as_string, type(number_as_string))
```

### Example: Calculate two numbers entered by the user

```python
first_number = float(input("Enter the first number: "))
second_number = float(input("Enter the second number: "))

print("Sum:", first_number + second_number)
print("Difference:", first_number - second_number)
print("Product:", first_number * second_number)
print("Quotient:", first_number / second_number)
```
