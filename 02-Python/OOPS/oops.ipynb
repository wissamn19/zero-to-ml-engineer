## 1. Python Classes

A **class** is a blueprint. It defines the data (attributes) and behavior (methods) that its objects will have.

```python
class Dog:
    # Class attribute: shared by all instances
    species = "Canis familiaris"

    # Constructor: runs when a new object is created
    def __init__(self, name, age):
        # Instance attributes: unique to each object
        self.name = name
        self.age = age

print(Dog.species)
print(Dog)
```

**Key points**
- `class` keyword defines a class. Names use `CamelCase`.
- `__init__` initializes each new object.
- `self` refers to the current instance.
- Class attributes are shared; instance attributes belong to one object.

```python
# Empty class with `pass`
class Placeholder:
    pass

# Class docstring
class Car:
    """Represents a car."""
    wheels = 4

print(Car.__doc__)
```

## 2. Python Objects

An **object** (instance) is a concrete thing created from a class.

```python
dog1 = Dog("Rex", 3)
dog2 = Dog("Luna", 5)

print(dog1.name, dog1.age)
print(dog2.name, dog2.age)
print(dog1.species, dog2.species)   # shared class attribute
```

```python
# Modify and add attributes
dog1.age = 4
dog1.color = "brown"     # dynamically added (only to dog1)
print(dog1.age, dog1.color)
print(hasattr(dog2, "color"))

# Delete an attribute / object
del dog1.color
print(hasattr(dog1, "color"))

# Identity vs equality
a = Dog("Max", 2)
b = Dog("Max", 2)
print(a == b)        # False: different objects
print(a is b)        # False
print(id(a), id(b))

# Inspect an object
print(type(a))
print(isinstance(a, Dog))
print(a.__dict__)
```

## 3. Python Methods

Methods are functions defined inside a class. Three kinds:
- **Instance methods**: take `self`, work on one object.
- **Class methods**: take `cls`, work on the class (`@classmethod`).
- **Static methods**: take neither, utility functions (`@staticmethod`).

```python
class Student:
    school = "BBA University"
    count = 0

    def __init__(self, name, grades=None):
        self.name = name
        self.grades = grades or []
        Student.count += 1

    # Instance method
    def add_grade(self, grade):
        self.grades.append(grade)

    def average(self):
        return sum(self.grades) / len(self.grades) if self.grades else 0

    # Class method
    @classmethod
    def from_string(cls, text):
        name, *grades = text.split(",")
        return cls(name, [float(g) for g in grades])

    @classmethod
    def total_students(cls):
        return cls.count

    # Static method
    @staticmethod
    def is_passing(grade):
        return grade >= 10

    # Special (dunder) methods
    def __str__(self):
        return f"Student({self.name}, avg={self.average():.2f})"

    def __repr__(self):
        return f"Student(name={self.name!r}, grades={self.grades!r})"

    def __len__(self):
        return len(self.grades)


s1 = Student("Wis")
s1.add_grade(15)
s1.add_grade(12)
print(s1)
print(repr(s1))
print(len(s1))

s2 = Student.from_string("Sara,14,16,11")
print(s2)
print(Student.total_students())
print(Student.is_passing(s2.average()))
```

## 4. Python Inheritance

A **child class** reuses and extends a **parent class**.

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return "..."

    def describe(self):
        return f"{self.name} says {self.speak()}"


class Dog(Animal):                 # single inheritance
    def speak(self):               # override
        return "Woof"

class Cat(Animal):
    def __init__(self, name, indoor=True):
        super().__init__(name)     # call parent constructor
        self.indoor = indoor

    def speak(self):
        return "Meow"

print(Dog("Rex").describe())
print(Cat("Tom").describe())
print(issubclass(Dog, Animal), isinstance(Cat("x"), Animal))
```

### Multi-level and multiple inheritance

```python
# Multi-level
class Puppy(Dog):
    def speak(self):
        return super().speak() + " (squeaky)"

print(Puppy("Bit").describe())

# Multiple inheritance
class Swimmer:
    def swim(self):
        return "Swimming"

class Walker:
    def walk(self):
        return "Walking"

class Duck(Animal, Swimmer, Walker):
    def speak(self):
        return "Quack"

d = Duck("Donald")
print(d.describe(), d.swim(), d.walk())

# Method Resolution Order
print(Duck.__mro__)
```

## 5. Python Polymorphism

Same interface, different behavior. Python supports it through:
1. Method overriding
2. Duck typing
3. Operator overloading
4. Built-in functions working on many types

```python
# 1. Overriding + common interface
class Shape:
    def area(self):
        raise NotImplementedError("Subclasses must implement area()")

class Rectangle(Shape):
    def __init__(self, w, h):
        self.w, self.h = w, h
    def area(self):
        return self.w * self.h

class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):
        return 3.14159 * self.r ** 2

shapes = [Rectangle(3, 4), Circle(2), Rectangle(5, 5)]
for s in shapes:
    print(type(s).__name__, round(s.area(), 2))
```

```python
# 2. Duck typing: "if it quacks like a duck..."
class Robot:
    def speak(self):
        return "Beep"

class Human:
    def speak(self):
        return "Hello"

for thing in (Robot(), Human(), Dog("Rex")):
    print(thing.speak())     # no shared parent needed

# 4. Built-ins are polymorphic
print(len("hello"), len([1, 2, 3]), len({"a": 1}))
print(1 + 2, "a" + "b", [1] + [2])
```

```python
# 3. Operator overloading
class Vector:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __mul__(self, k):
        return Vector(self.x * k, self.y * k)

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

v1, v2 = Vector(1, 2), Vector(3, 4)
print(v1 + v2)
print(v1 * 3)
print(v1 == Vector(1, 2))
```

## 6. Python Data Abstraction

Abstraction hides complexity and exposes only what's necessary. In Python, use the `abc` module: an **abstract class** cannot be instantiated and forces subclasses to implement certain methods.

```python
from abc import ABC, abstractmethod

class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        """Process a payment."""

    @abstractmethod
    def refund(self, amount):
        """Refund a payment."""

    # Concrete method shared by all subclasses
    def receipt(self, amount):
        return f"Receipt: {amount} DZD via {self.__class__.__name__}"


class CreditCard(PaymentMethod):
    def pay(self, amount):
        return f"Charged {amount} to card"
    def refund(self, amount):
        return f"Refunded {amount} to card"

class Cash(PaymentMethod):
    def pay(self, amount):
        return f"Received {amount} in cash"
    def refund(self, amount):
        return f"Returned {amount} in cash"


# PaymentMethod()  # TypeError: can't instantiate abstract class
for m in (CreditCard(), Cash()):
    print(m.pay(1000), "|", m.receipt(1000))
```

```python
# Missing an abstract method -> error
class Broken(PaymentMethod):
    def pay(self, amount):
        return "paid"

try:
    Broken()
except TypeError as e:
    print("Error:", e)
```

## 7. Python Encapsulation

Encapsulation bundles data with methods and controls access to it. Python uses naming conventions:

| Syntax | Meaning |
|---|---|
| `name` | public |
| `_name` | protected (convention: internal use) |
| `__name` | private (name-mangled) |

```python
class Account:
    def __init__(self, owner, balance):
        self.owner = owner          # public
        self._bank = "BEA"          # protected (convention)
        self.__balance = balance    # private (name-mangled)

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be positive")
        self.__balance += amount

    def withdraw(self, amount):
        if amount > self.__balance:
            raise ValueError("Insufficient funds")
        self.__balance -= amount

    def get_balance(self):
        return self.__balance


acc = Account("Wis", 1000)
acc.deposit(500)
print(acc.get_balance())
print(acc.owner, acc._bank)

try:
    print(acc.__balance)
except AttributeError as e:
    print("Error:", e)

# Name mangling: still reachable, but you shouldn't
print(acc._Account__balance)
```

### Properties: Pythonic getters and setters

```python
class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius        # goes through the setter

    @property
    def celsius(self):
        return self._celsius

    @celsius.setter
    def celsius(self, value):
        if value < -273.15:
            raise ValueError("Below absolute zero")
        self._celsius = value

    @property
    def fahrenheit(self):             # read-only computed property
        return self._celsius * 9 / 5 + 32


t = Temperature(25)
print(t.celsius, t.fahrenheit)
t.celsius = 30
print(t.fahrenheit)

try:
    t.celsius = -500
except ValueError as e:
    print("Error:", e)

try:
    t.fahrenheit = 100
except AttributeError as e:
    print("Error:", e)
```
