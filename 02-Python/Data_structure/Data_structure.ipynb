## 1. Python Strings and Built-in Functions

A string is a sequence of characters used to represent text.

````python
text = "Hello Python"

print(text)
print(text[0])
print(text[-1])
print(len(text))
````

### Common string methods

`upper()`, `lower()`, `capitalize()`, `title()`, `strip()`, `replace()`, `split()`, `join()`, `find()`, `count()`, `startswith()`, and `endswith()`.

````python
text = "  hello python world  "

print(text.upper())
print(text.lower())
print(text.strip())
print(text.replace("python", "programming"))
print(text.split())
print(text.find("python"))
print(text.count("o"))
print(text.strip().startswith("hello"))
print(text.strip().endswith("world"))
````

### f-strings

f-strings make it easy to insert variables into text.

````python
name = "Alex"
age = 21

print(f"My name is {name} and I am {age} years old.")
````

## 2. Python Lists and Boolean Variables

A list is an ordered, mutable collection. Lists use square brackets `[]`.

````python
fruits = ["apple", "banana", "orange"]

print(fruits)
print(fruits[0])

fruits[1] = "mango"
print(fruits)
````

### Common list methods

`append()`, `insert()`, `extend()`, `remove()`, `pop()`, `clear()`, `sort()`, `reverse()`, `index()`, and `count()`.

````python
numbers = [5, 2, 8, 2, 10]

numbers.append(20)
numbers.insert(1, 100)
numbers.extend([30, 40])
numbers.remove(2)

print(numbers)

last_item = numbers.pop()
print("Popped:", last_item)

numbers.sort()
print("Sorted:", numbers)

numbers.reverse()
print("Reversed:", numbers)

print("Index of 100:", numbers.index(100))
print("Count of 2:", numbers.count(2))
````

### Boolean variables

A Boolean has only two values: `True` and `False`.

Boolean values are commonly used in conditions.

````python
is_student = True
has_license = False

print(is_student)
print(type(is_student))

age = 20
is_adult = age >= 18

print(is_adult)
print(age >= 18 and has_license)
print(age < 18 or has_license)
print(not has_license)
````

## 3. Python Sets

A set is an unordered collection of unique values.

Sets are useful for removing duplicates and performing mathematical set operations.

````python
numbers = {1, 2, 3, 3, 4, 4, 5}

print(numbers)
print(len(numbers))

numbers.add(6)
numbers.update([7, 8])
numbers.discard(2)

print(numbers)
````

### Set operations

Useful methods include `union()`, `intersection()`, `difference()`, and `symmetric_difference()`.

````python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

print("Union:", set_a.union(set_b))
print("Intersection:", set_a.intersection(set_b))
print("Difference:", set_a.difference(set_b))
print("Symmetric difference:", set_a.symmetric_difference(set_b))
````

## 4. Python Dictionaries

A dictionary stores data as key-value pairs.

````python
student = {
    "name": "Alex",
    "age": 21,
    "grade": 85
}

print(student)
print(student["name"])

student["city"] = "Algiers"
student["age"] = 22

print(student)
````

### Common dictionary methods

`keys()`, `values()`, `items()`, `get()`, `update()`, `pop()`, `popitem()`, and `clear()`.

````python
student = {
    "name": "Alex",
    "age": 21,
    "grade": 85
}

print("Keys:", student.keys())
print("Values:", student.values())
print("Items:", student.items())
print("Name:", student.get("name"))
print("City:", student.get("city", "Not available"))

student.update({"city": "Algiers"})
removed_grade = student.pop("grade")

print("Removed:", removed_grade)
print(student)
````

### Looping through a dictionary

````python
student = {"name": "Alex", "age": 21, "grade": 85}

for key, value in student.items():
    print(key, ":", value)
````

## 5. Python Tuples

A tuple is an ordered collection that cannot be changed after creation.

Tuples use parentheses `()`.

````python
coordinates = (10, 20)

print(coordinates)
print(coordinates[0])
print(type(coordinates))

numbers = (10, 20, 10, 30, 10)

print(numbers.count(10))
print(numbers.index(30))
````

### Tuple unpacking

You can assign tuple values to multiple variables.

````python
person = ("Alex", 21, "Python")

name, age, language = person

print(name)
print(age)
print(language)
````

### List vs Tuple

- List: mutable, written with `[]`
- Tuple: immutable, written with `()`
- Lists are useful for data that changes.
- Tuples are useful for fixed data.


## 6. Functions in Python

A function is a reusable block of code. Functions are defined with `def`.

````python
def say_hello():
    print("Hello!")

say_hello()
say_hello()
````

### Parameters and return values

````python
def greet(name):
    print(f"Hello, {name}!")

greet("Alex")

def add(a, b):
    return a + b

result = add(10, 5)
print(result)
````

### Default parameters

````python
def greet(name="Guest"):
    print(f"Hello, {name}!")

greet("Alex")
greet()
````

### Example: reusable calculation

````python
def calculate_average(a, b, c):
    return (a + b + c) / 3

average = calculate_average(10, 15, 20)
print("Average:", average)
````

## 7. Lambda Functions

A lambda is a small anonymous function.

Syntax:

```python
lambda arguments: expression
```

````python
square = lambda x: x ** 2

print(square(5))
print(square(10))

add = lambda a, b: a + b
print(add(5, 3))
````

### Lambda with `sorted()`

Lambda functions are often useful for choosing the value used for sorting.

````python
students = [
    ("Alex", 85),
    ("Sarah", 92),
    ("John", 75)
]

students_sorted = sorted(students, key=lambda student: student[1])

print(students_sorted)
````

## 8. Iterators and Generators

An iterator provides values one at a time.

`iter()` creates an iterator and `next()` retrieves the next value.

````python
fruits = ["apple", "banana", "orange"]

fruit_iterator = iter(fruits)

print(next(fruit_iterator))
print(next(fruit_iterator))
print(next(fruit_iterator))
````

### Custom iterator

An iterator can implement `__iter__()` and `__next__()`.

````python
class CountUpTo:
    def __init__(self, maximum):
        self.maximum = maximum
        self.current = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= self.maximum:
            value = self.current
            self.current += 1
            return value
        raise StopIteration

counter = CountUpTo(3)

for number in counter:
    print(number)
````

### Generators

A generator is an easy way to create an iterator. It uses `yield` and produces values one at a time.

````python
def count_up_to(maximum):
    number = 1

    while number <= maximum:
        yield number
        number += 1

for number in count_up_to(5):
    print(number)
````

### Generator expression

````python
squares = (number ** 2 for number in range(1, 6))

for square in squares:
    print(square)
````

## 9. Exception Handling and Importing Libraries

Exception handling lets a program respond to runtime errors.

Main keywords:
- `try`
- `except`
- `else`
- `finally`


````python
try:
    number = int(input("Enter a number: "))
    print(10 / number)

except ValueError:
    print("Please enter a valid integer.")

except ZeroDivisionError:
    print("You cannot divide by zero.")
````

### `else` and `finally`

````python
try:
    number = int("10")
    result = 100 / number

except (ValueError, ZeroDivisionError):
    print("Something went wrong.")

else:
    print("Calculation successful:", result)

finally:
    print("This block always runs.")
````

### Raising an exception

Use `raise` when you want to create an exception yourself.

````python
age = 15

if age < 18:
    raise ValueError("Age must be 18 or older.")
````

### Importing libraries

Python has many built-in modules. Use `import` to use them.

````python
import math

print(math.sqrt(25))
print(math.pi)
print(math.ceil(4.2))
print(math.floor(4.8))
````
